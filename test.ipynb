```python
from PIL import Image, ImageDraw

# Load image
img = Image.open("Screenshot 2026-04-09 074517.png")
draw = ImageDraw.Draw(img)

# Get API response
result = response.json()

# Extract boxes
boxes = result["data"][0]["bounding_boxes"]

width, height = img.size

# Draw all categories
for category in boxes:
    for box in boxes[category]:
        x_min = box["x_min"] * width
        y_min = box["y_min"] * height
        x_max = box["x_max"] * width
        y_max = box["y_max"] * height

        draw.rectangle([x_min, y_min, x_max, y_max], outline="red", width=3)

# Show image
img.show()
```
