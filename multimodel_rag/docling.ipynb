```python
import json
import logging
import time
from pathlib import Path

from docling.datamodel.accelerator_options import AcceleratorDevice, AcceleratorOptions
from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import PdfPipelineOptions, TableStructureOptions
from docling.document_converter import DocumentConverter, PdfFormatOption

logging.basicConfig(level=logging.INFO)

# 🔹 Provide your PDF path here
input_doc_path = Path("C:\\Users\\pabbidi\\Downloads\\NV-Ingestion\\Data\\2310.02542v1.pdf")

# ---------------- Pipeline Configuration ----------------
pipeline_options = PdfPipelineOptions()

pipeline_options.do_ocr = True
pipeline_options.do_table_structure = True

pipeline_options.table_structure_options = TableStructureOptions(
    do_cell_matching=True
)

# OCR language (change if needed)
pipeline_options.ocr_options.lang = ["en"]

pipeline_options.accelerator_options = AcceleratorOptions(
    num_threads=4,
    device=AcceleratorDevice.AUTO
)

# ---------------- Document Converter ----------------
doc_converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
    }
)

# ---------------- Convert Document ----------------
start_time = time.time()

conv_result = doc_converter.convert(input_doc_path)

end_time = time.time() - start_time

print(f"Document converted in {end_time:.2f} seconds.")

# ---------------- Export Results ----------------
output_dir = Path("scratch")
output_dir.mkdir(parents=True, exist_ok=True)

doc_filename = conv_result.input.file.stem

# JSON
with open(output_dir / f"{doc_filename}.json", "w", encoding="utf-8") as fp:
    fp.write(json.dumps(conv_result.document.export_to_dict()))

# TXT
with open(output_dir / f"{doc_filename}.txt", "w", encoding="utf-8") as fp:
    fp.write(conv_result.document.export_to_markdown(strict_text=True))

# Markdown
with open(output_dir / f"{doc_filename}.md", "w", encoding="utf-8") as fp:
    fp.write(conv_result.document.export_to_markdown())

# DocTags
with open(output_dir / f"{doc_filename}.doctags", "w", encoding="utf-8") as fp:
    fp.write(conv_result.document.export_to_doctags())

print("Files saved in 'scratch' folder")
```

```python
import json
import logging
import time
from collections.abc import Iterable
from pathlib import Path

import yaml
from docling_core.types.doc import ImageRefMode

from docling.backend.docling_parse_backend import DoclingParseDocumentBackend
from docling.datamodel.base_models import ConversionStatus, InputFormat
from docling.datamodel.document import ConversionResult
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.document_converter import DocumentConverter, PdfFormatOption

logging.basicConfig(level=logging.INFO)
_log = logging.getLogger(__name__)

# ---------------- EXPORT CONFIG ----------------
USE_V2 = True
USE_LEGACY = False


# ---------------- EXPORT FUNCTION ----------------
def export_documents(conv_results: Iterable[ConversionResult], output_dir: Path):

    output_dir.mkdir(parents=True, exist_ok=True)

    success_count = 0
    failure_count = 0
    partial_success_count = 0

    for conv_res in conv_results:

        if conv_res.status == ConversionStatus.SUCCESS:

            success_count += 1
            doc_filename = conv_res.input.file.stem

            if USE_V2:

                conv_res.document.save_as_json(
                    output_dir / f"{doc_filename}.json",
                    image_mode=ImageRefMode.PLACEHOLDER,
                )

                conv_res.document.save_as_html(
                    output_dir / f"{doc_filename}.html",
                    image_mode=ImageRefMode.EMBEDDED,
                )

                conv_res.document.save_as_doctags(
                    output_dir / f"{doc_filename}.doctags.txt"
                )

                conv_res.document.save_as_markdown(
                    output_dir / f"{doc_filename}.md",
                    image_mode=ImageRefMode.PLACEHOLDER,
                )

                conv_res.document.save_as_markdown(
                    output_dir / f"{doc_filename}.txt",
                    image_mode=ImageRefMode.PLACEHOLDER,
                    strict_text=True,
                )

                # YAML Export
                with open(output_dir / f"{doc_filename}.yaml", "w") as fp:
                    fp.write(yaml.safe_dump(conv_res.document.export_to_dict()))

                # Markdown Export
                with open(output_dir / f"{doc_filename}.md", "w") as fp:
                    fp.write(conv_res.document.export_to_markdown())

                # Text Export
                with open(output_dir / f"{doc_filename}.txt", "w") as fp:
                    fp.write(conv_res.document.export_to_markdown(strict_text=True))


            if USE_LEGACY:

                with open(output_dir / f"{doc_filename}.legacy.json","w",encoding="utf-8") as fp:
                    fp.write(json.dumps(conv_res.legacy_document.export_to_dict()))

                with open(output_dir / f"{doc_filename}.legacy.txt","w",encoding="utf-8") as fp:
                    fp.write(conv_res.legacy_document.export_to_markdown(strict_text=True))

                with open(output_dir / f"{doc_filename}.legacy.md","w",encoding="utf-8") as fp:
                    fp.write(conv_res.legacy_document.export_to_markdown())

                with open(output_dir / f"{doc_filename}.legacy.doctags.txt","w",encoding="utf-8") as fp:
                    fp.write(conv_res.legacy_document.export_to_document_tokens())


        elif conv_res.status == ConversionStatus.PARTIAL_SUCCESS:

            _log.info(f"Document {conv_res.input.file} partially converted")

            for item in conv_res.errors:
                _log.info(item.error_message)

            partial_success_count += 1

        else:

            _log.info(f"Document {conv_res.input.file} failed")

            failure_count += 1


    _log.info(
        f"Processed {success_count + partial_success_count + failure_count} docs | "
        f"{failure_count} failed | "
        f"{partial_success_count} partial"
    )

    return success_count, partial_success_count, failure_count


# ---------------- INPUT FILES ----------------
# Replace with your PDFs

input_doc_paths = [
    Path("C:\\Users\\pabbidi\\Downloads\\Edge_01_compressed.pdf"),
    Path("C:\\Users\\pabbidi\\Downloads\\LLM_Assignment_03 24MAI0079.pdf"),
]


# ---------------- PIPELINE CONFIG ----------------

pipeline_options = PdfPipelineOptions()

pipeline_options.generate_page_images = False


# ---------------- DOC CONVERTER ----------------

doc_converter = DocumentConverter(

    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_options=pipeline_options,
            backend=DoclingParseDocumentBackend
        )
    }

)


# ---------------- RUN CONVERSION ----------------

start_time = time.time()

conv_results = doc_converter.convert_all(
    input_doc_paths,
    raises_on_error=False
)


# ---------------- EXPORT OUTPUT ----------------

success_count, partial_count, failure_count = export_documents(
    conv_results,
    output_dir=Path("Batch Conversion")
)

end_time = time.time() - start_time

print(f"\nConversion completed in {end_time:.2f} seconds")

print(f"Success: {success_count}")
print(f"Partial: {partial_count}")
print(f"Failed: {failure_count}")
```

```python
import json
import logging
from pathlib import Path
import yaml

from docling.backend.pypdfium2_backend import PyPdfiumDocumentBackend
from docling.datamodel.base_models import InputFormat
from docling.document_converter import (
    DocumentConverter,
    PdfFormatOption,
    WordFormatOption,
)
from docling.pipeline.simple_pipeline import SimplePipeline
from docling.pipeline.standard_pdf_pipeline import StandardPdfPipeline

logging.basicConfig(level=logging.INFO)
_log = logging.getLogger(__name__)

# ---------------- INPUT FILES ----------------
input_paths = [
    Path("C:\\Users\\pabbidi\\Downloads\\PROJECT_EXPLAINED.docx"),
    Path("C:\\Users\\pabbidi\\Downloads\\CoursePage-Material_WINSEM2024-25_MCSE505L_TH_VL2024250506461_18939_2025-04-27_10-41-54\\WINSEM2024-25_MCSE505L_TH_VL2024250506461_2025-04-02_Reference-Material-I.pptx"),
    Path("C:\\Users\\pabbidi\\Downloads\\download.jpg"),
    Path("C:\\Users\\pabbidi\\Downloads\\DL_ASS2_24MAI0087.pdf"),
]

# ---------------- DOCUMENT CONVERTER ----------------
doc_converter = DocumentConverter(
    allowed_formats=[
        InputFormat.PDF,
        InputFormat.IMAGE,
        InputFormat.DOCX,
        InputFormat.HTML,
        InputFormat.PPTX,
        InputFormat.ASCIIDOC,
        InputFormat.CSV,
        InputFormat.MD,
    ],
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_cls=StandardPdfPipeline,
            backend=PyPdfiumDocumentBackend
        ),
        InputFormat.DOCX: WordFormatOption(
            pipeline_cls=SimplePipeline
        ),
    },
)

# ---------------- CONVERT DOCUMENTS ----------------
conv_results = doc_converter.convert_all(input_paths)

# ---------------- OUTPUT DIRECTORY ----------------
out_path = Path("MultiFormat")
out_path.mkdir(exist_ok=True)

# ---------------- EXPORT RESULTS ----------------
for res in conv_results:

    print(
        f"Document {res.input.file.name} converted."
        f"\nSaved markdown output to: {out_path}"
    )

    _log.debug(res.document._export_to_indented_text(max_text_len=16))

    with (out_path / f"{res.input.file.stem}.md").open("w") as fp:
        fp.write(res.document.export_to_markdown())

    with (out_path / f"{res.input.file.stem}.json").open("w") as fp:
        fp.write(json.dumps(res.document.export_to_dict()))

    with (out_path / f"{res.input.file.stem}.yaml").open("w") as fp:
        fp.write(yaml.safe_dump(res.document.export_to_dict()))

print("All documents converted successfully.")
```
